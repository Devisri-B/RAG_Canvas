```python
from canvasapi import Canvas
from dotenv import load_dotenv
import os

load_dotenv()
canvas = Canvas(os.getenv("CANVAS_BASE_URL"), os.getenv("CANVAS_API_TOKEN"))

courses = canvas.get_courses()
for course in courses:
    print(f"ID: {course.id} | Name: {course.name}")
```

```python
# Get your specific course
course = canvas.get_course(2)  # use the ID of course

# get files in the course
files = course.get_files()
for f in files:
    print(f.id, f.display_name, f.url)
```

```python
from dotenv import dotenv_values
v = dotenv_values("/Users/devisri/Desktop/Research/Deepsikha/RAG/canvas-self-hosted/.env")
c = Canvas(v["CANVAS_BASE_URL"].strip(), v["CANVAS_API_TOKEN"].strip())
course = c.get_course(2)
for m in course.get_modules():
    print(f"[Module {m.id}] name={m.name!r} unlock_at={getattr(m,'unlock_at',None)} published={getattr(m,'published',None)}")
    items = list(m.get_module_items())
    if not items:
        print("    (no items yet)")
    for it in items:
        print(f"    - type={it.type} title={getattr(it,'title','')!r} content_id={getattr(it,'content_id',None)}")
```
